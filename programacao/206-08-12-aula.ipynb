```python
numeros =  [1,5,1,3,7,9,4,6,10,15]

maximo = 0
minimo = 0
for n in numeros:
    if maximo < n:
        maximo = n
    if minimo > n:
        minimo = n

print(f"O valor maximo é: {maximo}, o valor minimo é: {minimo}")
```

```python
notas = [7,8,5,9,6]

soma = 0

for nota in notas:
    soma += nota

media = soma / len(notas)

print(len(notas))
print(f"A média é: {media}")
```

```python
quantidade = int(input("Qual a quantidade de alunos: "))

soma = 0
for i in range(quantidade):
    nome = input("Qual o nome do aluno: ")
    nota = float(input("Qual a nota: "))

    if nota >= 6:
        print(f"O aluno {nome}, está Aprovado")
    else:
        print(f"O aluno {nome}, está Reprovado")

    soma += nota

media = soma / quantidade

print(f"A média da sala é: {media}")

    
```
