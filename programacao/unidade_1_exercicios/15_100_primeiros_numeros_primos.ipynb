#codigo desenvolvido para listar os 100 primeiros numeros primos 

```python
def verifica_numero_primo(qtde_primo):
    contador_de_primo = 0
    numero = 3
    while contador_de_primo < qtde_primo:
        iterador = 2
        resto = 0
        while iterador < numero:
            # print(f"iterador: { iterador}")

            resto = numero % iterador

            if resto == 0:
                eh_primo = False
                break
            else: 
                eh_primo = True

                if iterador + iterador > numero: #caso o iterador passe da metade do numero de entrada, não tem mais como dividir
                    break
                else:
                    iterador += 1
        if eh_primo:
            print(f"O número {numero} é Primo")
            contador_de_primo += 1
        numero += 1
```

```python

verifica_numero_primo(100)
```
