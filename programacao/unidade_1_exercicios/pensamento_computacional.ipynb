## Código base

```python
n = int(input("Digite um número: "))

if n <= 1:
    print("Número inválido.")
else:
    divisores = 0
    soma = 0

    for i in range(1, n + 1):
        if n % i == 0:
            divisores += 1
            if i != n:
                soma += i

    if divisores == 2:
        print("Número primo")
    else:
        print("Não é primo")

    if soma == n:
        print("Número perfeito")
    else:
        print("Não é perfeito")
```

```python
# 1 - Explicando como cada pilar do pensamento computacional foi aplicado no código.

n = int(input("Digite um número: "))



#o pilar da ABSTRAÇÃO do pensamento computacional é aplicado no decorrer do código, apresentando as entradas com input(), o processamento com o laço "for" e as saídas com print().
if n <= 1:
    print("Número inválido.")
else:
    divisores = 0
    soma = 0

# O pilar da ABSTRAÇÃO do pensamento computacional é aplicado nessa parte do código focando apenas na lógica de verificar os divisoes e a soma dos denominadores. 
    for i in range(1, n + 1): #O pilar de reconhecimento de padrões do pensamento computacional é aplicado nessa parte do código, na identificação de que o laço "for" atende tanto a condição de número primo quanto número perfeito.
        if n % i == 0:
            divisores += 1
            if i != n:
                soma += i

#O pilar da DECOMPOSIÇÃO do pensamento computacional é aplicado ao dividir o problema em parte (superior e inferior) A parte de cima realiza o calculo e a parte de baixo imprime.
    if divisores == 2:
        print("Número primo")
    else:
        print("Não é primo")

    if soma == n:
        print("Número perfeito")
    else:
        print("Não é perfeito")
```
