#Esse código foi desenvolvido para verificar se uma entrada é palindromo

```python
i = int(input("Qual o número a ser invertido: "))
valor_origem = i
modulo = ""
while i != 0:
    modulo += str(i % 10)
    print(modulo)
    i = i // 10

if valor_origem == int(modulo):
    print(f"O número {valor_origem} é um palíndromo.")
else:
    print(f"O número {valor_origem} não é um palíndromo.")
```
