#Esse código foi desenvolvido para listar os iterador primeiros números perfeitos, sendo perfeito aqueles que a soma de seus divisores dá ele mesmo

```python
def lista_iterador_perfeito(numero):

    j = numero // 2
    resultado = 0 


    while j >= 1:
        if numero % j == 0:
            resultado += j
            # print(resultado)

        j -= 1

    if resultado == numero:
        return 1
    else:
        return 0
    # else:
    #     print("não é número perfeito")
```

```python
qtde = 0
numero = 0
while qtde < 5:
    if lista_iterador_perfeito((numero)) == 1:
        qtde += 1
        print(f"O número {numero} é perfeito")
    numero += 1

# print(qtde)
```
