#Esse Código foi desenvolvido para contar numeros pares e impares, sendo o 0 a condição de saída do programa

```python
i = 1
qtde_par = 0
qtde_impar = 0


while i != 0 :
    i = int(input("Digite o número: "))
    if i % 2 == 0:
        print("é par")
        qtde_par += 1
    else:
        print("é impar")
        qtde_impar += 1

print(f"Tivemos ao todo {qtde_par} números pares e {qtde_impar} números impares")
```
