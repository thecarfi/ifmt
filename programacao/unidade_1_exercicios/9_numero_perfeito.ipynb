#esse código foi desenvolvido para identificar quais números são perfeitos, sendo perfeito a soma de todos os seus divisores da ele mesmo

```python
x = int(input("Qual o número:"))

j = x // 2
resultado = 0 


while j >= 1:
    if x % j == 0:
        resultado += j
        print(resultado)

    j -= 1

if resultado == x:
    print("é número perfeito")
else:
    print("não é número perfeito")

```
