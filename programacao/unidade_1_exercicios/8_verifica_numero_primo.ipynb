# código criado para verificador se um input() é um número primo

```python
numero = int(input("digite o número: "))
iterador = 2
resto = 0
while iterador < numero:
    # print(f"iterador: { iterador}")

    resto = numero % iterador

    if resto == 0:
        eh_primo = False
        break
    else: 
        eh_primo = True

        if iterador + iterador > numero: #caso o iterador passe da metade do numero de entrada, não tem mais como dividir
            break
        else:
            iterador += 1

if eh_primo:
    print(f"O número é Primo")
```
