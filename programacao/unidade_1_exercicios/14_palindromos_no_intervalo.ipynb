#esse código foi desenvolvido para apresentar o palíndromos no intervalo

```python
def Verifica_se_eh_palindromo(i):
    valor_origem = i
    modulo = ""
    while i != 0:
        modulo += str(i % 10)
        # print(modulo)
        i = i // 10

        if valor_origem == int(modulo):
            print(f"O número {valor_origem} é um palíndromo.")
        # else:
            # print(f"O número {valor_origem} não é um palíndromo.")
```

```python
i = 1
while i <=  1000:
    Verifica_se_eh_palindromo(i)
    i += 1
```
