#esse código foi desenvolviod para soma de série numérica

```python
numero = int(input("Digite o número: "))

iterador = 2
resultado = 1
while iterador <= numero:
    resultado += 1 / iterador
    print(f"resultado parcial: {1 / iterador}")

    iterador += 1

print(f"O resultado é: {resultado:.2f}")
```
