# Esse código foi desenvolvido para calcular o valor fatoriall de um determinado número passado no input()


```python
i =  int(input("Digite o numero para realizar o calculo fatorial: ")) 
resultado = 1
j = i
while j != 0:
    resultado *= i
    print(resultado)
    j -= 1

print(f"O resultado fatorial de {i} é: {resultado}")
```
