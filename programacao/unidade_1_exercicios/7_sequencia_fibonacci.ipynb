#esse código foi desenvolvido para gerar X quantidade da sequencia de fibonacci, sendo X o valor do input()

```python
i = int(input("Digite o número para gerar a sequência de Fibonacci:"))

j = 1
resultado = [0]

while j < i:
    if i == 1:
        j += 1
        break
    elif j == 1:
        resultado.append(1)
        j += 1
    else:      
        resultado.append(resultado[j-2] + resultado[j-1])
        j += 1

if i == 0:
    print("Para listar a sequência de Fibonacci, o número deve ser maior que 0.")
else:
    print(f"A sequência de Fibonacci até {i} é: {resultado}")
print("Fim da execução")
```
