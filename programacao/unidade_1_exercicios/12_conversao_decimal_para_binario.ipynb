#esse código foi desenvolvido para converter um número decimal em binario

```python
numero = int(input("digite o valor: "))

numero_binario = ""

while numero > 1:
    print(f"numero: {numero}")
    numero_binario = str(numero % 2) + numero_binario
    numero = numero // 2




numero_binario = "1" + numero_binario
print(f"binario: {numero_binario}")
```
