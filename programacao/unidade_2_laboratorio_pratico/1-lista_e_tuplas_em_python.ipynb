### Exercício 1
Crie uma lista com 5 cidades de Mato Grosso. Depois:
1. exiba a primeira e a última;
2. adicione uma nova cidade;
3. remova uma cidade;
4. ordene alfabeticamente;
5. mostre a quantidade de elementos.

```python
# SUA RESPOSTA


cidades = ["CUIABA","VARZEA GRANDE","SINOP","CACERES","CHAPADA DOS GUIMARAES"]

print(cidades[0]) #1 - Exibindo a primeira cidade da lista
print(cidades[-1]) #2 - Exibindo a última cidade da lista

cidades.remove("VARZEA GRANDE") #3 - Removendo uma cidade

cidades.sort() #4 - Ordenando as cidades alfabeticamente
print(cidades)

print(f"A lista tem: {len(cidades)} registros") #5 - Imprimindo a quantidade de elementos da lista
```

## 3. Percorrendo listas

```python
notas = [7.5, 8.0, 6.5, 9.0]

for nota in notas:
    print(nota)

media = sum(notas) / len(notas)
print("Média:", media)
```

### Exercício 2
Dada a lista abaixo, mostre apenas os valores maiores ou iguais a 7.

```python
notas = [5.5, 8.0, 6.0, 9.5, 7.0, 4.5]

# SUA RESPOSTA

for nota in notas:
    if nota >= 7:
        print(f"Nota: {nota}")
```

### Exercício 3
Crie uma tupla com os dias da semana e:
- mostre o terceiro dia;
- mostre os cinco primeiros;
- percorra todos com `for`.

```python
# SUA RESPOSTA
dias_da_semana = ("Domingo","Segunda-feira","Terça-feira","Quarta-feira","Quinta-feira","Sexta-feira","Sabado")
print(dias_da_semana[2]) #imprimindo o terceiro dia
print(dias_da_semana[0:5]) #imprimindo os cincos primeiros dias da tupla

for dia in dias_da_semana: #Percorrendo todos os dias com FOR e imprimindo cada um
    print(dia)
```

## Desafio final
Crie uma lista de estudantes. Para cada estudante, armazene uma tupla com:
`(nome, nota1, nota2)`.

Depois:
1. calcule a média de cada estudante;
2. informe se está aprovado (`média >= 7`);
3. apresente o estudante com a maior média.

```python
estudantes = [
    ("Ana", 8.0, 7.5),
    ("Bruno", 6.0, 5.5),
    ("Carla", 9.0, 8.5),
    ("Diego", 7.0, 7.0),
]

# SUA SOLUÇÃO

medias = []
for estudante in estudantes:
    print(f"Nome do estudante: {estudante[0]}")
    media = estudante[1] + estudante[2] / 2

    media_aluno = (estudante[0], media) #Criando nova tupla do estudante
    medias.append(media_aluno) #adicionando nova tupla do estudante a uma lista


maior_nota = 0
nome_aluno_maior_nota = ""

for nota in medias:
    if nota[1] > maior_nota:
        maior_nota = nota[1]
        nome_aluno_maior_nota = nota[0]

print(medias)
print(f"O aluno de maior nota é {nome_aluno_maior_nota} com a nota {maior_nota}")
```

## Conclusão
Registre em uma célula de texto: **quando você escolheria uma lista e quando escolheria uma tupla?**

A escolha entre uma lista e uma tupla está na necessidade posterior de realizar uma edição desse dado ou não.

Por exemplo, caso venha criar uma lista com o cpf de um aluno, esse seria um dado que não altera ao longo do tempo, logo, se fosse para escolher entre tupla e lista, a opção correta seria tupla, para não ocorrer o risco de haver alterações.

Outro exemplo seria a nota de um aluno em uma prova. Pode haver uma reconsideração de um questão da prova, oque geraria a necessidade de alterar a nota dos alunos. Tal necessidade faz com que optamos pela escolha de uma lista, visto que, pode haver a necessidade de alterar o dado de nota dessa lista.
