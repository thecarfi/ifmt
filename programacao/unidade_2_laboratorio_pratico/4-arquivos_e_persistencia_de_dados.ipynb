## 1. Criando um arquivo

```python
texto = '''Ana,8.5
Bruno,6.2
Carla,9.1
Diego,7.0
'''

with open("notas.txt", "w", encoding="utf-8") as arquivo:
        arquivo.write(texto)

print("Arquivo criado.")
```

## 2. Lendo o arquivo

```python
with open("notas.txt", "r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()

print(conteudo)
```

### Exercício 1
Leia o arquivo linha a linha e mostre somente os nomes.

```python
# SUA RESPOSTA

with open("notas.txt", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        nome = linha.strip().split(",")[0] #O strip() remove o \n e o split() remove a virgula, convertendo o conteúdo em lista, posteirormente pegando a posição 0
        print(nome)
```

## 3. Trabalhando com CSV

```python
import csv

dados = [
    ["nome", "idade", "cidade"],
    ["Ana", 22, "Cuiabá"],
    ["Bruno", 25, "Várzea Grande"],
    ["Carla", 27, "Rondonópolis"],
]
# print(dados)
with open("pessoas.csv", "w", newline="", encoding="utf-8") as arquivo:
    writer = csv.writer(arquivo)
    writer.writerows(dados)

print("CSV gerado.")
```

## 4. Lendo CSV

```python
import csv

with open("pessoas.csv", "r", encoding="utf-8") as arquivo:
    reader = csv.DictReader(arquivo)
    pessoas = list(reader)

print(pessoas)
```

### Exercício 2
Mostre apenas as pessoas com idade maior ou igual a 25.

```python
# SUA RESPOSTA

import csv

with open("pessoas.csv","r", encoding="utf-8") as arquivo: #Abre o arquivo
    reader = csv.DictReader(arquivo)#Cria um leitor do arquivo com o nome reader
    pessoas = list(reader) #converte o reader em uma lista

    for pessoa in pessoas: #itera a lista procurando a que tem idade >= 25
        idade = int(pessoa["idade"])
        if idade >= 25:
            print(pessoa)
```

## Desafio final — Relatório
Leia `pessoas.csv` e gere um arquivo `relatorio.txt` contendo:
- total de registros;
- maior idade;
- menor idade;
- média das idades;
- lista de cidades diferentes.

```python
# SUA SOLUÇÃO

import csv
maior_idade = 0
menor_idade = 300
contador = 0
media_idade = 0
cidades  = set()

#lendo arquivo pessoas
with open("pessoas.csv", "r",encoding="utf-8") as arquivo:
    reader = csv.DictReader(arquivo)
    pessoas = list(reader)

    for pessoa in pessoas: #Pegando a maior e menor idade
        contador += 1 #Serve como contador para média e para contar registros
        idade = int(pessoa["idade"])
        media_idade += idade

        if idade > maior_idade:
            maior_idade = idade

        if idade < menor_idade:
            menor_idade = idade
        cidades.add(pessoa["cidade"])

    media_idade = int(media_idade / contador)

# print(f"A maior idade: {maior_idade}")
# print(f"A menor idade: {menor_idade}")
# print(f"A média idade: {media_idade}")
# print(cidades)

# print(pessoas)
# print(len(pessoas))



texto = f'''Total registros,{contador}
maior idade,{maior_idade}
menor idade,{menor_idade}
média das idades,{media_idade}
lista de cidades diferentes,{cidades}
'''
# print(texto)


with open("relatorio.txt", "w",encoding="utf-8") as arquivo:
    arquivo.write(texto)


with open("relatorio.txt","r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()


print(conteudo)
```
