## 1. Transformação com `for`

```python
numeros = [1, 2, 3, 4, 5]
quadrados = []

for n in numeros:
    quadrados.append(n ** 2)

print(quadrados)
```

## 2. A mesma solução com list comprehension

```python
numeros = [1, 2, 3, 4, 5]
quadrados = [n ** 2 for n in numeros]
print(quadrados)
```

## 3. Filtragem

```python
numeros = list(range(1, 21))
pares = [n for n in numeros if n % 2 == 0]
print(pares)
```

### Exercício 2
A partir da lista de notas, crie uma lista contendo apenas notas >= 7.

```python
notas = [4.5, 7.0, 8.2, 5.9, 9.1, 6.8, 7.5]


# SUA RESPOSTA
notas_na_media = [n for n in notas if n >= 7]
print(notas_na_media)
```

## 4. Lambda

```python
dobro = lambda x: x * 2
print(dobro(5))
```

## 5. `map`, `filter` e `sorted`

```python
valores = [10, 3, 7, 1, 9]

print(list(map(lambda x: x * 2, valores)))
print(list(filter(lambda x: x >= 7, valores)))
print(sorted(valores, reverse=True))
```

### Exercício 3
Ordene os estudantes abaixo pela média, da maior para a menor,
utilizando `sorted(..., key=lambda ...)`.

```python
estudantes = [
    {"nome":"Bruno", "media":6.4},
    {"nome":"Carla", "media":9.3},
    {"nome":"Ana", "media":8.5},
    {"nome":"Diego", "media":7.7},
]

# SUA RESPOSTA
print(sorted(estudantes, key=lambda a: a["media"], reverse=True))
```
