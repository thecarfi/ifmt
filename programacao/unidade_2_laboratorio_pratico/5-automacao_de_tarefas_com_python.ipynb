```python
import csv

dados = [
    ["id", "tipo", "bairro", "prioridade", "status"],
    [1, "Iluminação", "Centro", "Alta", "Aberta"],
    [2, "Trânsito", "Cristo Rei", "Média", "Concluída"],
    [3, "Segurança", "Centro", "Alta", "Aberta"],
    [4, "Iluminação", "Jardim Glória", "Baixa", "Concluída"],
    [5, "Trânsito", "Centro", "Alta", "Aberta"],
    [6, "Iluminação", "Cristo Rei", "Média", "Aberta"],
]

with open("solicitacoes.csv", "w", newline="", encoding="utf-8") as arquivo:
    writer = csv.writer(arquivo)
    writer.writerows(dados)

print("Base criada.")
```

## 1. Ler automaticamente os dados

```python
import csv

with open("solicitacoes.csv", "r", encoding="utf-8") as arquivo:
    registros = list(csv.DictReader(arquivo))

print("Quantidade:", len(registros))
print(registros[:2])
```

### Exercício 1
Calcule:
- total de solicitações;
- total abertas;
- total concluídas;
- quantidade de prioridade Alta.

```python
# SUA RESPOSTA


import csv

with open("solicitacoes.csv","r",encoding="utf-8") as arquivo:
    registros = list(csv.DictReader(arquivo))

    qtde_registros = sum(1 for linha in registros)
    qtde_registro_aberto = sum(1 for linha in registros if linha["status"] == "Aberta")
    qtde_registro_concluido = sum(1 for linha in registros if linha["status"] == "Concluída")
    qtde_reigstros_prioridade_alta = sum(1 for linha in registros if linha["prioridade"] == "Alta")
print(f"A quantidade de registros é: {qtde_registros}")
print(f"A quantidade de registros com status aberto é: {qtde_registro_aberto}")
print(f"A quantidade de registros com status Concuído é {qtde_registro_concluido}")
print(f"A quantidade de registros com prioridade alta é: {qtde_reigstros_prioridade_alta}")
```

## 2. Indicadores por categoria
Gere um dicionário com a quantidade de registros por tipo.

```python
por_tipo = {}

for r in registros:
    print(r)
    tipo = r["tipo"]
    por_tipo[tipo] = por_tipo.get(tipo, 0) + 1

print(por_tipo)
```

### Exercício 2
Gere também:
- quantidade por bairro;
- conjunto de bairros diferentes;
- lista apenas das solicitações abertas e de prioridade Alta.

```python
# SUA RESPOSTA

por_bairro = {}

for registro in registros:
    bairro = registro["bairro"]
    por_bairro[bairro] = por_bairro.get(bairro,0)+1

print(por_bairro)
```

```python
from datetime import datetime

agora = datetime.now().strftime("%d/%m/%Y %H:%M:%S")
with open("relatorio_automacao.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("RELATÓRIO DE SOLICITAÇÕES\n")
    arquivo.write("=" * 40 + "\n")
    arquivo.write(f"Gerado em: {agora}\n")
    arquivo.write(f"Total de registros: {len(registros)}\n")
    arquivo.write("\nQuantidade por tipo:\n")
    for tipo, qtd in por_tipo.items():
        arquivo.write(f"- {tipo}: {qtd}\n")

print("Relatório gerado.")


with open("relatorio_automacao.txt", "r", encoding="utf-8") as arquivo:
    dados = list(csv.DictReader(arquivo))
    for linha in dados:
        print(linha)
# print(registros)
```
