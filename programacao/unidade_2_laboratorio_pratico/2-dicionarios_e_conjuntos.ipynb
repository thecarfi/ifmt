## 3. Lista de dicionários

```python
alunos = [
    {"nome":"Ana", "media":8.5},
    {"nome":"Bruno", "media":6.2},
    {"nome":"Carla", "media":9.1},
    {"nome":"Diego", "media":7.0},
]

for aluno in alunos:
    situacao = "Aprovado" if aluno["media"] >= 7 else "Reprovado"
    print(aluno["nome"], "-", situacao)
```

### Exercício 2
Calcule:
- média geral da turma;
- maior média;
- nomes dos estudantes aprovados.

```python
# SUA RESPOSTA
notas = 0
maior_media = 0
for aluno in alunos:
    notas += aluno["media"]
    maior_media = aluno["media"] if aluno["media"] > maior_media else maior_media


print(f"Média geral da turma = {round(notas / len(alunos),2)}")

print(f"A maior média da sala é: {maior_media}")

lista = [aluno["nome"] for aluno in alunos if aluno["media"] >= 7]
print(f"Alunos aprovados: {lista}")
```

## 4. Conjuntos

```python
turma_a = {"Ana", "Bruno", "Carla", "Diego"}
turma_b = {"Carla", "Diego", "Eva", "Fernando"}

print("União:", turma_a | turma_b)
print("Interseção:", turma_a & turma_b)
print("Somente A:", turma_a - turma_b)
```

## Desafio final — Cadastro de ocorrências
Considere uma lista de ocorrências, cada uma representada por um dicionário
com `tipo`, `bairro` e `prioridade`.

Produza:
1. quantidade total de ocorrências;
2. conjunto de bairros diferentes;
3. quantidade por tipo;
4. apenas ocorrências de prioridade "Alta".

```python
ocorrencias = [
    {"tipo":"Trânsito", "bairro":"Centro", "prioridade":"Alta"},
    {"tipo":"Iluminação", "bairro":"Cristo Rei", "prioridade":"Média"},
    {"tipo":"Trânsito", "bairro":"Centro", "prioridade":"Baixa"},
    {"tipo":"Segurança", "bairro":"Jardim Glória", "prioridade":"Alta"},
    {"tipo":"Iluminação", "bairro":"Centro", "prioridade":"Alta"},
]

# SUA SOLUÇÃO


bairros = set()
tipos = {}
for ocorrencia in ocorrencias:
    bairro  = ocorrencia["bairro"]
    tipo  = ocorrencia["tipo"]
    
    bairros.add(bairro)

    if tipo not in tipos:
        tipos[tipo] = {"qtde": 1, "prioridade": ocorrencia["prioridade"]}
    else:
        tipos[tipo]["qtde"] += 1


print(tipos)
print(f"Quantidade ocorrências: {len(ocorrencias)}") #1 - Lista a quantidade de ocorrências
print(f"Quantidade de bairros: {len(bairros)}") #2 - Lista a quantidade de bairros


for chave in tipos:
    valor = tipos[chave]["qtde"]
    print(f"O tipo: {chave} tem: {valor} ocorrências {'é prioridade Alta' if tipos[chave]['prioridade'] == 'Alta' else ''}") #3 e 4 - Listar a quantidade de registro por tipo de ocorrências e informa se é prioridade alta
```
