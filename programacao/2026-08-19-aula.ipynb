# Para verificar se o número é perfeito, sendo perfeito aquele se elevar ao expoente seus números individualmente em um loop chegará a 1.

```python
numero = int(input("Qual o número: "))

eh_feliz = False
lista = []
j = 0
lista.append(numero)
while j < 20:
    resultado = 0
    for i in str(numero):
        resultado += int(i)**2
    numero = resultado
    print(f"O resultado: {resultado}")
    if resultado == 1:
        eh_feliz = True 
        break
    elif resultado in lista:
        break
    lista.append(resultado)
    # print(f"O número é: {numero}")
    print(f"A lista é: {lista}")
    j += 1


if eh_feliz:
    print(f"O número {lista[0]} é Feliz")
else:
    print(f"O número {lista[0]} Não é Feliz")
```
